# **Dictionary Method**

```python
nums= [1,1,1,3,3,4,3,2,4,2]
```

```python
counts={}
def contains_duplicate(arr):
    for i in arr:
        counts[i]= counts.get(i,0)+1
        if counts.get(i,0)>1:
            return True
    return False
```

```python
contains_duplicate(nums)
```

# **Set Method**

```python
nums= [1,1,1,3,3,4,3,2,4,2]
```

```python
my_set = set()
def contains_duplicate(arr):
    for i in arr:
        my_set.add(i)
    if len(my_set) != len(nums):
        return True
    else:
        return False

contains_duplicate(nums)
```
